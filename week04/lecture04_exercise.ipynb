```python
import pandas as pd
import plotly.express as px


# Dataset: Gapminder — GDP, Life Expectancy, Population by Country
# Source: Gapminder Foundation (gapminder.org)

df = px.data.gapminder()
print(f"Loaded: {len(df)} rows")
print(df.head())
```

```python
# explore

print(df.info())
print("Years:", sorted(df['year'].unique()))
print("Continents:", df['continent'].unique())
print(df.describe().round(1))
```

## Task 1 — Scatter: life expectancy change over time

**What to build:** A scatter showing **GDP per capita vs life expectancy** for **two years** (2002 and 2007) to show how both moved — use **colour for year** (just 2 colours), **one continent only**.

Choose any continent except Africa (that was the example). Highlight the change direction.

> 💡 Filter: `df.loc[df['continent'] == 'YOUR_CHOICE']` then filter years


```python
# Task 1
# YOUR CODE HERE
import plotly.express as px
import plotly.graph_objects as go

continent = "Europe"
years = [2002, 2007]
plot_df = df.loc[(df["continent"] == continent) & (df["year"].isin(years))].copy()

fig = px.scatter(
    plot_df,
    x="gdpPercap",
    y="lifeExp",
    color="year",
    color_discrete_map={2002: "#7A9ECA", 2007: "#D95F59"},
    hover_name="country",
    hover_data={"gdpPercap": ":,.0f", "lifeExp": ":.1f", "year": True},
    title="Europe, 2002–2007: most countries gained life expectancy as incomes rose",
    labels={
        "gdpPercap": "GDP per capita (US$)",
        "lifeExp": "Life expectancy (years)",
        "year": "Year",
    },
)

# Connect each country's 2002 and 2007 positions to make the direction of change visible.
for country, country_df in plot_df.groupby("country"):
    country_df = country_df.sort_values("year")
    if len(country_df) == 2:
        p0, p1 = country_df.iloc[0], country_df.iloc[1]
        fig.add_trace(go.Scatter(
            x=[p0["gdpPercap"], p1["gdpPercap"]],
            y=[p0["lifeExp"], p1["lifeExp"]],
            mode="lines",
            line=dict(color="rgba(100,100,100,0.25)", width=1),
            showlegend=False,
            hoverinfo="skip",
        ))

fig.update_layout(template="plotly_white", legend_title_text="Year")
fig.update_xaxes(rangemode="tozero")
fig.show()
```

## Task 2 — Bubble chart: tell a story

**What to build:** A bubble chart (full 2007 dataset, all countries) where:
- x = GDP per capita (log scale)
- y = life expectancy
- size = population
- colour = ONE continent highlighted (your choice), all others grey
- At least one annotation explaining the highlighted group's story

> This is the grey-and-highlight technique applied to a bubble chart.


```python
# Task 2import plotly.express as px

year_df = df.loc[df["year"] == 2007].copy()
highlight_continent = "Asia"
year_df["group"] = year_df["continent"].apply(
    lambda c: highlight_continent if c == highlight_continent else "Other continents"
)

fig = px.scatter(
    year_df,
    x="gdpPercap",
    y="lifeExp",
    size="pop",
    color="group",
    color_discrete_map={
        "Asia": "#D95F59",
        "Other continents": "#C8CDD2",
    },
    size_max=42,
    hover_name="country",
    hover_data={
        "continent": True,
        "gdpPercap": ":,.0f",
        "lifeExp": ":.1f",
        "pop": ":,",
        "group": False,
    },
    title="Asia in 2007: wide income gaps, but life expectancy spans nearly the full global range",
    labels={
        "gdpPercap": "GDP per capita (US$, log scale)",
        "lifeExp": "Life expectancy (years)",
        "pop": "Population",
        "group": "Group",
    },
)

fig.update_xaxes(type="log")
fig.update_layout(template="plotly_white", legend_title_text="")
fig.add_annotation(
    x=1800,
    y=72,
    xref="x",
    yref="y",
    text="Asia contains both lower-income<br>countries and high-income economies",
    showarrow=True,
    arrowhead=2,
    ax=75,
    ay=-55,
    bgcolor="white",
    bordercolor="#D95F59",
    borderwidth=1,
)
fig.show()
# YOUR CODE HERE
```
